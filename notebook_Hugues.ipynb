```python
%matplotlib inline
import matplotlib.pyplot as plt
```

### Etage

#### location

#### détermination des outliers

```python
df = df_loc_68
```

```python
feature = "etage"
```

```python
df[feature].value_counts()
```

#### je garde jusqu'à max 26 ème étage

```python
max_feature = 26
```

```python
df_max = df[df[feature] <= max_feature]
```

```python
plt.hist(df_max[feature], bins =max_feature)
plt.title (f"repartition des biens à la location en fonction de {feature} du bien");
```

```python
df_max.plot.scatter(x=feature, y='prix_bien', c="g");
```
